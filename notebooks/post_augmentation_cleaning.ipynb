```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
pth = os.path.abspath(os.path.join(os.getcwd(), ".."))
pth = os.path.abspath(os.path.join(pth, "data/augmented_raw"))
print(pth)  # Affiche le répertoire parent
```

```python
df_train_augmented = pd.read_csv(pth + "/train_augmented.csv")
df_val_augmented = pd.read_csv(pth + "/val_augmented.csv")
df_test_augmented = pd.read_csv(pth + "/test_augmented.csv")
```

```python
# Distribution des labels pour l'ensemble d'entraînement
df_train_augmented.label.value_counts().plot(kind='bar')
```

```python
# Distribution des labels pour l'ensemble de validation
df_val_augmented.label.value_counts().plot(kind='bar')
```

```python
# Distribution des labels pour l'ensemble de test
df_test_augmented.label.value_counts().plot(kind='bar')
```

```python
# Est-ce qu'il y a des valeurs manquantes  dans les données d'entraineemnt?
print(df_train_augmented.isnull().sum())
```

```python
# Est ce qu'il ya des doublons de textes ? 
print(df_train_augmented['text'].duplicated().sum())
```

```python
# supprimer les doublons
df_train_augmented.drop_duplicates(subset='text', keep='first', inplace=True)
```

```python
# Est-ce qu'il y a des valeurs manquantes ?
print(df_val_augmented.isnull().sum())
```

```python
# Est ce qu'il ya des doublons de textes ? 
print(df_val_augmented['text'].duplicated().sum())
```

```python
# supprimer les doublons
df_val_augmented.drop_duplicates(subset='text', keep='first', inplace=True)
```

```python
# Est-ce qu'il y a des valeurs manquantes ?
print(df_test_augmented.isnull().sum())
```

```python
# Est ce qu'il ya des doublons de text
print(df_test_augmented['text'].duplicated().sum())
```

```python
# supprimer les doublons
df_test_augmented.drop_duplicates(subset='text', keep='first', inplace=True)
```

```python
# Vérifier que tous les labels sont présents dans les tous les sets de données
print(set(df_train_augmented.label.unique()) == set(df_val_augmented.label.unique()) == set(df_test_augmented.label.unique()))
```

```python
pth = os.path.abspath(os.path.join(os.getcwd(), ".."))
pth = os.path.abspath(os.path.join(pth, "data/augmented"))
print(pth)  # Affiche le répertoire parent
```

```python
df_train_augmented.to_csv(pth + "/train.csv", index=False)
df_val_augmented.to_csv(pth + "/val.csv", index=False)
df_test_augmented.to_csv(pth + "/test.csv", index=False)
```
